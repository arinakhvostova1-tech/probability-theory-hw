## Тема 4. Условная вероятность

> Решить классически и смоделировать решение задачи. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

```python
# Классическое решение
p = 0

for k in range(4):
    p_transfer = sc.comb(15, k) * sc.comb(5, 3-k) / sc.comb(20, 3)
    p_white = (6 + k) / 10
    p += p_transfer * p_white

print(f"Ответ: {p}")
```

```python
# Моделирование
N = 100000

success = 0

for _ in range(N):
    urn1 = np.array([1] * 15 + [0] * 5)
    transferred = np.random.choice(urn1, size=3, replace=False)

    urn2 = np.array([1] * 6 + [0])
    urn2 = np.concatenate([urn2, transferred])

    ball = np.random.choice(urn2)

    if ball == 1:
        success += 1

print(f"Ответ: {success / N}")
```

### Задача 2.9
> В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 1) не будут проданы 5 пакетов; 2) будет продано: а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) наивероятнейшее число пакетов. 

```python
n = 9
p = 0.2
result_1 = binom.pmf(5, n, (1-p))
print(f"1) Ответ: {round(result_1, 4)}") 

result_2a = binom.cdf(1, n, p)
print(f"2a) Ответ: {round(result_2a, 4)}") 

result_2b = binom.cdf(2, n, p)
print(f"2б) Ответ: {round(result_2b, 4)}")

result_2c = 1 - binom.cdf(1, n, p)
print(f"2в) Ответ: {round(result_2c, 4)}")

result_2d = 0
k0 = [k for k in range(n+1) if n*p - (1-p) <= k <= n*p + p]
for i in k0:
    result_2d += binom.pmf(i, n, p)
print(f"2г) Ответ: {round(result_2d, 4)}")
```

### Задача 2.11
> По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины:
а) 480 предприятий; б) наивероятнейшее число предприятий; в) не менее 480; г) от 480 до 520. 

```python
n = 1000
p = 0.5

s = sqrt(n*p*(1-p))
x = (480 - n*p) / s
result_a = norm.pdf(x) / s
print(f"a) Ответ: {round(result_a, 4)}") 

result_b = 0
k0 = [k for k in range(n+1) if n*p - (1-p) <= k <= n*p + p]
for i in k0:
    x = (i - n*p)/s
    result_b += norm.pdf(x) / s
print(f"б) Ответ: {round(result_b, 4)}") 

mu = n*p
normal_dist = norm(loc=mu, scale=s)

result_c = normal_dist.cdf(n) - normal_dist.cdf(480)
result_c_p = normal_dist.cdf(n + 0.5) - normal_dist.cdf(480 - 0.5)
print(f"в) Ответ без поправки: {round(result_c, 4)} \nОтвет с поправкой: {round(result_c_p, 4)}") 

result_d = normal_dist.cdf(520) - normal_dist.cdf(480)
result_d_p = normal_dist.cdf(520 + 0.5) - normal_dist.cdf(480 - 0.5)
print(f"г) Ответ без поправки: {round(result_d, 4)} \nОтвет с поправкой: {round(result_d_p, 4)}") 
```
