### Задача 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

```python
outcomes = list(it.product(range(1, 7), repeat=3))
print(f"а) Ответ: {sum(sum(x) == 11 for x in outcomes) / len(outcomes)}")
```

```python
print(f"б) Ответ: {sum(sum(x) > 10 for x in outcomes) / len(outcomes)}")
```

## Тема 4. Условная вероятность

> Решить классически и смоделировать решение задачи. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

```python
# Классическое решение
p = 0

for k in range(4):
    p_transfer = sc.comb(15, k) * sc.comb(5, 3-k) / sc.comb(20, 3)
    p_white = (6 + k) / 10
    p += p_transfer * p_white

print(f"Ответ: {p}")
```

```python
# Моделирование
N = 100000

success = 0

for _ in range(N):
    urn1 = np.array([1] * 15 + [0] * 5)
    transferred = np.random.choice(urn1, size=3, replace=False)

    urn2 = np.array([1] * 6 + [0])
    urn2 = np.concatenate([urn2, transferred])

    ball = np.random.choice(urn2)

    if ball == 1:
        success += 1

print(f"Ответ: {success / N}")
```
